```python
customers = spark.table("silver_customers")
products = spark.table("silver_products")
orders = spark.table("silver_orders")
```

```python
sales = (
    orders
    .join(products, "product_id")
    .join(customers, "customer_id")
)
```

```python
from pyspark.sql.functions import sum

customer_sales = (
    sales
    .groupBy(
        "customer_id",
        "name",
        "city"
    )
    .agg(
        sum("price").alias("total_sales")
    )
)
```

```python
customer_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_customer_sales")
```

```python
from pyspark.sql.functions import count

top_products = (
    sales
    .groupBy(
        "product_name"
    )
    .agg(
        count("*").alias("orders")
    )
)
```

```python
top_products.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_top_products")
```

```python
from pyspark.sql.functions import sum

city_sales = (
    sales
    .groupBy("city")
    .agg(
        sum("price").alias("sales")
    )
)
```

```python
city_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_city_sales")
```
